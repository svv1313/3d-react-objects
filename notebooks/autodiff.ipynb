```python
import torch
import tensorflow as tf
```

```python
x = torch.tensor(5.0)
```

```python
x
```

```python
x.requires_grad_()
```

```python
y = x**2
```

```python
y.backward()
```

```python
x.grad
```

```python
x = tf.Variable(5.0)
```

```python
with tf.GradientTape() as t:
  t.watch(x)
  y = x**2
```

```python
t.gradient(y, x)
```
